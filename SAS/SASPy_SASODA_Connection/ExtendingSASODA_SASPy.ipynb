#### 4. Configure SASPy for ODA

You need to modify SASPy’s connection settings:

a. Find the config file path:

```
import saspy, os
print(saspy.__file__.replace('__init__.py','sascfg_personal.py'))

```

 Alternative syntax

```
import saspy
saspy.SAScfg
```

b. Edit sascfg_personal.py:

* Define ODA connection settings, including your Java path and ODA region hosts.

Example snippet (adjust for your system/region):

```
SAS_config_names = ['oda']
oda = {
    'java'   : 'C:\\Path\\To\\Java\\java.exe',  # or /usr/bin/java on macOS/Linux
    'iomhost': ['odaws01-usw2.oda.sas.com','odaws02-usw2.oda.sas.com'],
    'iomport': 8591,
    'authkey': 'oda',
    'encoding':'utf-8'
}

```

#### 5. Create Your Credentials File

* Create a plain-text file for ODA login credentials named _authinfo (Windows) or .authinfo (macOS/Linux) in your home directory.

* Store the "_authinfo" file in your Windows machine, typically "C:\Users\your-userid", for SAS clients like saspy, with the leading underscore being intentional. To get this to work, you will need a valid USERID and PW for SAS OnDemand for Academic in your "_authinfo" file. 

```
oda user YOUR_ODA_EMAIL_OR_USERNAME password YOUR_ODA_PASSWORD
```

* This file lets SASPy authenticate with ODA without prompting every time.

#### 6. Three .jar Files

Get the three ".jar" files from https://support.sas.com/downloads/package.htm?pid=2494#. Store those ".jar" files in the "saspy/java/iomclient" directory of the SASPy installation in your Python environment.

#### 7. Edit the "hosts" File

You may need to edit your "hosts" file (with no extension) in the folder "C:\Windows\System32\drivers\etc", as documented in the SAS Knowledge Base. Consider adding the following line to the "hosts" file only if needed.
```
127.0.0.1 localhost view-localhost

```

#### 8. Connect to SAS from Python

In a Python or Jupyter session:

```
import saspy
sas = saspy.SASsession()
print(sas)
```
If successful, SASPy will start a SAS session on the remote ODA server.

#### 9. Run SAS Code in Jupyter Notebooks

Once connected, you can submit SAS code directly in Python notebooks using SASPy’s magic commands:
```
from saspy.sas_magic import SASMagic
SASMagic.saslib = sas
%load_ext saspy.sas_magic
```

Then in a notebook cell:
```
%%SAS
proc print data=sashelp.class(obs=5);
run;
```

The above code runs and shows the output in JupyterLab.
