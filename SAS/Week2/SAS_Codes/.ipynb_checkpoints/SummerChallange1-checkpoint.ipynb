```sas
/* Program 1 */
/* Step 1: Reading the text file */
filename elements "c:\Users\pmuhuri\SASCourse\Week2\Week2data\TheElements.txt";
data lyrics;
infile elements pad;
input line $200.;
run;

```

```sas

/* Step 2: Converting all text to lowercase */
data lyrics_lower;
set lyrics;
line_lower = lowcase(line);
run;


```

```sas
/* Step 3: Counting occurrences of the word 'and' */
data count_and;
set lyrics_lower;
* and_count = countw(prxchange('s/\band\b/and/i', -1, line_lower), 'and');
and_count = countw(lowcase(line), 'and', ' ');
run;
proc print data=count_and;
run;
```

```sas
proc sql;
select sum(and_count) as total_and_count
from count_and;
quit;
```

```sas
proc sql;
	create table and_count as
	select word, count(word," and ") as count_and,
	count(word, "And ") as count_and_capital
	from Work.Elements;
	
    select sum(count_and)+sum(count_and_capital) as total_ands
	from and_count;
quit;
```

```sas
/* Program 3 */
/* Step 1: Reading the text file */
filename elements "c:\Users\pmuhuri\SASCourse\Week2\Week2data\TheElements.txt";
data lyrics;
infile elements;
input line $200.;
run;

/* Step 2: Converting all text to lowercase */
data lyrics_lower;
set lyrics;
line_lower = lowcase(line);
run;

/* Step 3: Counting occurrences of the word 'and' */
data count_and;
set lyrics_lower;

and_count = countw(prxchange('s/\band\b/and/i', -1, line_lower), 'and');
run;


proc sql;
select sum(and_count) as total_and_count
from count_and;
quit;
```
