
#### Week 2, Part 7: Listing files in a directory using R from a SAS PROC IML step

##### Check whether RLANG enables SAS to execute R language statements.

```sas
ods html close;
options nodate nonumber nonotes nosource;
proc options option=RLANG;
run;
```

```sas
PROC IML;
SUBMIT / R;
setwd ("C:/Explore/SAS/Week2/SAS_Codes")
list.files(pattern=".sas$", 
           full.names = TRUE, 
           ignore.case = TRUE)
ENDSUBMIT;
QUIT;
```

```sas
PROC IML;
SUBMIT / R;
setwd ("C:/Explore/SAS/Week2/SAS_Codes")
Sys.glob("*.sas")  # returns a sorted list of files
ENDSUBMIT;
QUIT;
```

```sas
proc iml;
submit / R;
dir <- "C:/Explore/SAS/Week1/SAS_Codes"
if (!dir.exists(dir)) {
  stop("Directory does not exist: ", dir)
}
files <- list.files(dir,
                     pattern="\\.sas$",
                     full.names=TRUE,
                     ignore.case=TRUE)
print(files)
endsubmit;
quit;
```
