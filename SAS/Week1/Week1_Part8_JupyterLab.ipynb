#### Week 1, Part 8: Launching JupyterLab

Method 1: Launch JupyterLab from the desired folder (recommended)
Open Anaconda Prompt (or Command Prompt if Jupyter is on your PATH).
Change to the folder containing your notebook:

```
cd "D:\Projects\SAS\Week1"
```

Method 2: Start JupyterLab in a specific folder from anywhere
You can specify the working directory directly:

```
jupyter lab --notebook-dir="D:\Projects\SAS\Week1"
```

Method 3: Open an individual notebook directly
You can also specify the notebook file:

```
jupyter lab "D:\Projects\SAS\Week1\Week1_Part1.ipynb"
```
