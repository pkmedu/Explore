#### Zainab's questions and my responses

Zainab emailed me on May 5, 2026, 2:04 PM, requesting clarification about the use of %NRBQUOTE.

In her email, she said, she expected the following code to output STAT4197&grad, since %NRBQUOTE should mask the macro trigger (&) embedded in &Course: 

```
%let grad = 6197;
%let Course = STAT4197&grad;
%put %nrbquote(&Course);
```
However, ... it outputs STAT41976197 instead. 

Her second question: If %NRBQUOTE isn't masking the & trigger, then what differentiates it from %BQUOTE?



Below are my responses.

Let me first run Zainab's code.

````sas
/* Zainab's code - Let us call it Variant 1 */
ods html close;
option nocenter nonotes nodate nosource nonumber;
%let grad = 6197;
%let Course = STAT4197&grad;
%put %nrbquote(&Course);
````

In the above code (**Variant 1**), `%put %nrbquote(&Course);` does NOT protect the macro trigger (&). So SAS normally resolves `&Course` to `STAT41976197`. 

Remember that %NRBQUOTE is an execution-time function that works on text values (including & and %) returned by the macro processor code execution.



**However, if your goal is to protect the macro trigger (&), use `%put %nrstr(&Course);` instead, as shown below  (Variant 2).**

````sas
/* Variant 2 */
ods html close;
option nocenter nonotes nodate nosource nonumber;
%let grad = 6197;
%let Course = STAT4197&grad;
%put %nrstr(&Course);
````

In the above code (**Variant 2**), `%NRSTR` operates during compile time: it prevents macro resolution and treats the text as literal, so `&Course` is not expanded.

Remember that `%NRSTR` is a compile-time function that works on constant text (i.e., that masks special characters and mnemonics, including & and %) passed to the macro processor.


**Variant 3 (An alternative to Zainab's original code): If you want `&grad` to resolve normally, you don't need to use any of the macro quoting functions, as shown below.**

````sas
/* Variant 3 */
ods html close;
option nocenter nonotes nodate nosource nonumber;
%let grad = 6197;
%let Course = STAT4197&grad;
%put &Course;
````

##### Difference between %BQUOTE and %NRBQUOTE

* %BQUOTE masks everything at macro execution time and treats (&) normally.

* When compared with %BQUOTE, the only difference is that the %NRBQUOTE function will resolve down as far as it can, but does not resolve any ampersands or percent signs in the final result.
