#### Macro Basics

* The SAS macro facility is a text substitution facility for reducing the amount
of text that you must enter to do common tasks.

* The macro facility has two components:

    * Macro Processor: the portion of Base SAS that does the macro work
    * Macro language: the syntax that communicates with the macro processor

#### Macro Triggers (& and %)


The ampersand (&) followed by a name references a macro variable:

* &dsn
* &group
* &year


The percent sign (%) followed by a name references a macro program,
a macro statement, or a macro function.

Macro name
* %dsn
* %group
* %year

Macro statement
* %do year = 2010 %to 2020;

Macro function
* %substr(&date,6,4)

#### When to quote a macro variable reference in SAS

A macro variable value is neither character nor numeric. 
It is just text. SAS decides what type it is based 
on the context in which it is used.  In the example below, 
x is a numeric variable, and y is a character variable.  

```sas
ods _all_ close;                  /* closes listing, html, rtf, etc. */
ods html5 style=htmlblue;         /* open fresh HTML destination */
options nodate nonumber notes nosource;
%let year=2018;
data test;
 x=&year; /* numeric assignment */
 y="&year"; /* character assignment */
run;
```

What happens:
* x = &year;
* Resolves to x = 2018;
* x is numeric because no quotes → numeric literal.
* y = "&year";
* Resolves to y = "2018";
* y is character because quotes are used.

DICTIONARY.COLUMNS (part of the DICTIONARY tables family) is a read-only table in SAS that contains information about all variables for all datasets currently accessible in your session. In the code snippet below, we use a SQL query to extract the metadata of the TEST dataset.

```sas
proc sql; 
 select varnum, name, type 
	from dictionary.columns 
	where libname = 'WORK' and memname = 'TEST'; 
quit;
```

As an alternative, in the code below, we use the SASHELP view VCOLUMN to query variable metadata for the TEST dataset.

```sas
proc print data=sashelp.vcolumn;
   var name type length;
   where upcase(libname)='WORK' and memname = 'TEST';
run;
```
