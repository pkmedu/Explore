#### DICTIONARY.TABLES 

[See more about DICTIONARY.TABLES here in SAS Documentation](https://documentation.sas.com/doc/en/pgmsascdc/9.4_3.5/sqlproc/n02s19q65mw08gn140bwfdh7spx7.htm#n00p9ntyv3an0yn12f2tx377nb9u)

* is a read-only object and contains information about known tables (i.e., SAS datasets).
* contains information about datasets ... allocated with the LIBNAME. The Table also includes LIBNAMEs typically allocated during initialization (SASUSER, SASHELP, and maps).

[SESUG Paper AT-279-2019 by Frank C. DiIorio](https://www.lexjansen.com/sesug/2019/SESUG2019_Paper-279_Final_PDF.pdf)

##### Create a macro variable with a list of SAS dataset names from DICTIONARY.TABLES where LIBNAME = "SASHELP" using PROC SQL 

```sas
ods html close;
options nosource nonumber nodate nonotes nosource;
proc sql noprint;
    select memname
        INTO :dslist separated by ' '
        FROM DICTIONARY.TABLES
        where libname = "SASHELP";
    %let ds = &sqlobs;
quit;

%put Number of SAS datasets = &ds;
%put ;
%put &=dslist;
%put;
```

##### Create a macro variable containing a horizontal list of countries  from a SASHELP SAS dataset

```sas
ods html close;
options nosource nonumber nodate nonotes nosource;
proc sql noprint;
    select distinct trim(name)
        INTO :clist separated by ','
        FROM SASHELP.demographics;        
quit;

%put Number of countries for all world regions combined = &sqlobs;
%put ;
%put &=clist;
%put;
```

##### Create a macro variable containing a horizontal list of countries under the region AFR

```sas
options nosource nonumber nodate nonotes nosource;
ods html close;
proc sql noprint;
    select distinct trim(name)
        INTO :AFR_clist separated by ','
        FROM SASHELP.demographics
        where region="AFR";
quit;

%put Number of countries for AFR_clist = &sqlobs;
%put ;
%put &=AFR_clist;
%put;
%put NOTE: End of region AFR processing;
%put ;
```

##### Macro approach 1 (macro with a positional parameter)

Create six macro variables, each containing a horizontal list of countries 
under each of the six regions.

```sas
ods html close;
options nosource nonumber nodate nonotes;
%macro runit(rname);
proc sql noprint;
    select distinct trim(name)
        INTO :&rname._clist separated by ','
        FROM SASHELP.demographics
        where region="&rname";
quit;
%put NOTE: Processing region &rname using a macro with positional parameter;
%put &rname._clist = %superq(&rname._clist);
%put Number of countries for &rname._clist = &sqlobs;
%put NOTE: End of region &rname processing;
%put;
%mend runit;
%runit(AFR)
%runit(AMR)
%runit(EMR)
%runit(SEAR)
%runit(WPR)
%runit(EUR)
```

##### Macro approach 2 

* The macro dynamically creates a macro variable named AFR_clist, AMR_clist, etc.
* %superq is used to safely display macro variable values without triggering issues with commas or special characters.
* %put prints debug info (macro variable content, SQL observation count) to the SAS log.

```sas
options nosource nonumber nodate nonotes;
%macro process_region(rname);
    proc sql noprint;
        select distinct trim(name)
            into :&rname._clist separated by ","
            from SASHELP.demographics
            where region="&rname";
    quit;
    %put NOTE: Processing region &rname using a macro Loop;
    %put NOTE: SQL Observation count = &sqlobs;
    %put &rname._clist = %superq(&rname._clist);
    %put NOTE: End of region &rname processing;
    %put;
%mend process_region;
%macro run_all;
    %let regions = AFR AMR EMR SEAR WPR EUR;
    %do i = 1 %to %sysfunc(countw(&regions));
        %let region = %scan(&regions, &i);
        %process_region(&region)
    %end;
%mend run_all;
%run_all
```

 ##### Macro to define dynamic macro variable creation and logging

```sas
options nosource nonumber nodate nonotes;
%macro new_process_region(rname);
    proc sql noprint;
        select distinct trim(name)
            into :&rname._clist separated by ","
            from SASHELP.demographics
            where region="&rname";
    quit;
    %put NOTE: Processing region &rname using a macro Loop;
    %put NOTE: SQL Observation count = &sqlobs;
    %put &rname._clist = %superq(&rname._clist);
    %put NOTE: End of region &rname processing;
    %put;
%mend new_process_region;
%new_process_region(AFR);
```

##### Macro approach 3 (macro call with CALL EXECUTE)

* CALL EXECUTE queues macro calls to execute after the DATA step completes.

* %nrstr ensures that the macro call is not resolved immediately; it will execute later with region substituted. 

* Each iteration generates a call like:

%new_process_region(AFR);
%new_process_region(AMR);
...

```sas
options nosource nonumber nodate nonotes;
data _null_;
    length region $4;
    array regions[6] $4 ('AFR' 'AMR' 'EMR' 'SEAR' 'WPR' 'EUR');
    
    do i = 1 to dim(regions);
        region = regions[i];
        call execute('%nrstr(%new_process_region(' || region || '));');
    end;
run;
```
